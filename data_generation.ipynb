### Generating 20 Stores

```python

store_locations = [
    ("Delhi", "North"),
    ("Gurugram", "North"),
    ("Bengaluru", "South"),
    ("Hyderabad", "South"),
    ("Mumbai", "West"),
    ("Pune", "West"),
    ("Kolkata", "East")
]


stores_data = []

store_number = 1

for city, region in store_locations:
    
    if city in ["Delhi", "Gurugram", "Bengaluru", "Hyderabad", "Mumbai", "Pune"]:
        num_stores = 3
    else:
        num_stores = 2

    for i in range(num_stores):
        store_id = f"S{store_number:03d}"
        store_name = f"{city} Store {i + 1}"

        store_type = np.random.choice(
            ["Mall", "High Street", "Standalone"]
        )

        store_size = np.random.randint(2500, 7001)

        opening_date = pd.Timestamp(
            np.random.choice(
                pd.date_range("2019-01-01", "2024-12-31")
            )
        ).date()

        stores_data.append({
            "store_id": store_id,
            "store_name": store_name,
            "city": city,
            "region": region,
            "store_type": store_type,
            "store_size_sqft": store_size,
            "opening_date": opening_date
        })

        store_number += 1


stores_df = pd.DataFrame(stores_data)

print(f"Number of stores: {len(stores_df)}")
stores_df
```

```python
# Defining product catalog

product_catalog = {
    "Clothing": {
        "T-Shirts": ["Basic Cotton Tee", "Premium Polo", "Graphic T-Shirt"],
        "Jeans": ["Slim Fit Jeans", "Regular Fit Jeans", "Relaxed Jeans"],
        "Jackets": ["Denim Jacket", "Bomber Jacket", "Lightweight Jacket"]
    },
    "Footwear": {
        "Sports Shoes": ["Running Shoes", "Training Shoes", "Walking Shoes"],
        "Casual Shoes": ["Canvas Sneakers", "Casual Loafers", "Slip-On Shoes"],
        "Sandals": ["Sports Sandals", "Casual Sandals"]
    },
    "Sportswear": {
        "Activewear": ["Training T-Shirt", "Performance Shorts", "Track Pants"],
        "Gym Wear": ["Gym Vest", "Compression Tights", "Workout Shorts"],
        "Outdoor": ["Hiking T-Shirt", "Outdoor Jacket"]
    },
    "Accessories": {
        "Bags": ["Backpack", "Laptop Bag", "Sling Bag"],
        "Watches": ["Classic Watch", "Sports Watch"],
        "Belts": ["Leather Belt", "Casual Belt"],
        "Wallets": ["Leather Wallet", "Card Holder"]
    },
    "Beauty & Personal Care": {
        "Skincare": ["Face Wash", "Moisturizer", "Sunscreen"],
        "Haircare": ["Shampoo", "Conditioner", "Hair Serum"]
    }
}

# Generated products
products_data = []
product_number = 1

for category, subcategories in product_catalog.items():
    for subcategory, product_names in subcategories.items():
        for product_name in product_names:
            
            # Randomly generate cost and selling price
            unit_cost = np.random.randint(300, 2501)
            
            # Selling price is higher than cost
            markup = np.random.uniform(1.3, 2.2)
            selling_price = round(unit_cost * markup, 2)

            products_data.append({
                "product_id": f"P{product_number:03d}",
                "product_name": product_name,
                "category": category,
                "subcategory": subcategory,
                "unit_cost": unit_cost,
                "selling_price": selling_price
            })

            product_number += 1

products_df = pd.DataFrame(products_data)

print(f"Number of products: {len(products_df)}")
products_df.head(10)
```

### Genrating Promotional Campaigns

```python
promotion_campaigns = [
    ("Festival Sale", 10),
    ("Weekend Sale", 5),
    ("Seasonal Sale", 15),
    ("Clearance Sale", 20),
    ("Special Offer", 10)
]

promotions_data = []

promotion_number = 1

for store_id in stores_df["store_id"]:

    # Each store gets 5–10 promotions
    num_promotions = np.random.randint(5, 11)

    for _ in range(num_promotions):

        product_id = np.random.choice(products_df["product_id"])

        campaign_type, discount = promotion_campaigns[
            np.random.randint(len(promotion_campaigns))
        ]

        # Random promotion start date
        start_date = pd.Timestamp(
            np.random.choice(
                pd.date_range("2026-01-01", "2026-06-15")
            )
        )

        # Promotion lasts 3–14 days
        duration = np.random.randint(3, 15)
        end_date = start_date + pd.Timedelta(days=duration)

        # Keep promotions within our six-month period
        if end_date > pd.Timestamp("2026-06-30"):
            end_date = pd.Timestamp("2026-06-30")

        promotions_data.append({
            "promotion_id": f"PR{promotion_number:04d}",
            "store_id": store_id,
            "product_id": product_id,
            "start_date": start_date,
            "end_date": end_date,
            "discount_percentage": discount,
            "campaign_type": campaign_type
        })

        promotion_number += 1

promotions_df = pd.DataFrame(promotions_data)

print(f"Number of promotions: {len(promotions_df)}")
promotions_df.head(10)
```

### SALES GENERATION — DEMAND FACTORS

```python
# Product popularity
# Higher value = product is more likely to be purchased
product_demand_weight = np.random.uniform(
    0.5, 2.0, len(products_df)
)

products_df["demand_weight"] = product_demand_weight


# Category seasonality
# Values represent relative demand by season.
# 1.0 = normal demand

seasonality = {
    "Winter": {
        "Clothing": 1.15,
        "Footwear": 1.00,
        "Sportswear": 0.95,
        "Accessories": 1.00,
        "Beauty & Personal Care": 0.95
    },

    "Spring": {
        "Clothing": 1.05,
        "Footwear": 1.05,
        "Sportswear": 1.05,
        "Accessories": 1.00,
        "Beauty & Personal Care": 1.05
    },

    "Summer": {
        "Clothing": 1.15,
        "Footwear": 1.10,
        "Sportswear": 1.15,
        "Accessories": 1.05,
        "Beauty & Personal Care": 1.20
    }
}


# Customer segment spending behavior
customer_segment_multiplier = {
    "Premium": 1.30,
    "Regular": 1.00,
    "Occasional": 0.80,
    "New": 0.90
}


# Store type demand multiplier
store_type_multiplier = {
    "Mall": 1.20,
    "High Street": 1.00,
    "Standalone": 0.85
}


# Day-of-week multiplier
day_multiplier = {
    0: 0.90,  # Monday
    1: 0.95,  # Tuesday
    2: 0.95,  # Wednesday
    3: 1.00,  # Thursday
    4: 1.05,  # Friday
    5: 1.20,  # Saturday
    6: 1.15   # Sunday
}


print("Demand factors configured successfully.")
```

### SALES GENERATION — TRANSACTIONS

```python
sales_data = []

# Create a date range for our six-month period
date_range = pd.date_range(
    "2026-01-01",
    "2026-06-30"
)

# Base number of transaction opportunities per day
base_daily_transactions = 280


for sale_date in date_range:

    # ----------------------------------------
    # 1. Determine the season
    # ----------------------------------------

    if sale_date.month <= 2:
        season = "Winter"
    elif sale_date.month == 3:
        season = "Spring"
    else:
        season = "Summer"


    # ----------------------------------------
    # 2. Day-of-week effect
    # ----------------------------------------

    daily_multiplier = day_multiplier[sale_date.dayofweek]


    # ----------------------------------------
    # 3. Holiday / event effect
    # ----------------------------------------

    event_multiplier = 1.0

    if sale_date.month == 1 and sale_date.day == 26:
        event_multiplier = 1.35

    elif sale_date.month == 3 and sale_date.day == 4:
        event_multiplier = 1.40

    elif sale_date.month == 4 and sale_date.day == 1:
        event_multiplier = 1.25


    # ----------------------------------------
    # 4. Number of transactions for the day
    # ----------------------------------------

    expected_transactions = int(
        base_daily_transactions
        * daily_multiplier
        * event_multiplier
    )

    # Add natural variation
    num_transactions = max(
        1,
        int(
            np.random.normal(
                expected_transactions,
                expected_transactions * 0.08
            )
        )
    )


    # ----------------------------------------
    # 5. Generate transactions
    # ----------------------------------------

    for _ in range(num_transactions):

        # Select a store
        store = stores_df.iloc[
            np.random.randint(len(stores_df))
        ]

        # Store effect
        store_multiplier = store_type_multiplier[
            store["store_type"]
        ]

        # Larger stores have more demand
        size_multiplier = (
            store["store_size_sqft"] / 4500
        )

        # Select a customer
        customer = customers_df.iloc[
            np.random.randint(len(customers_df))
        ]

        customer_multiplier = customer_segment_multiplier[
            customer["customer_segment"]
        ]


        # Select a product
        product = products_df.iloc[
            np.random.randint(len(products_df))
        ]

        # Product popularity
        product_multiplier = product["demand_weight"]


        # ----------------------------------------
        # 6. Seasonal category effect
        # ----------------------------------------

        seasonal_multiplier = seasonality[
            season
        ][product["category"]]


        # ----------------------------------------
        # 7. Promotion effect
        # ----------------------------------------

        active_promotion = promotions_df[
            (promotions_df["store_id"] == store["store_id"]) &
            (promotions_df["product_id"] == product["product_id"]) &
            (promotions_df["start_date"] <= sale_date) &
            (promotions_df["end_date"] >= sale_date)
        ]

        promotion_multiplier = 1.0
        discount = 0.0

        if not active_promotion.empty:

            promotion = active_promotion.iloc[0]

            discount = promotion[
                "discount_percentage"
            ]

            # Promotion increases demand
            promotion_multiplier = (
                1 + discount / 100 * 0.50
            )


        # ----------------------------------------
        # 8. Combine demand factors
        # ----------------------------------------

        demand_score = (
            store_multiplier
            * size_multiplier
            * customer_multiplier
            * product_multiplier
            * seasonal_multiplier
            * promotion_multiplier
        )

        # Natural randomness
        demand_score *= np.random.uniform(
            0.85,
            1.15
        )


        # ----------------------------------------
        # 9. Decide whether purchase happens
        # ----------------------------------------

        purchase_probability = min(
            0.95,
            0.30 * demand_score
        )

        if np.random.random() > purchase_probability:
            continue


        # ----------------------------------------
        # 10. Quantity purchased
        # ----------------------------------------

        quantity = np.random.choice(
            [1, 2, 3],
            p=[0.75, 0.20, 0.05]
        )


        # ----------------------------------------
        # 11. Revenue
        # ----------------------------------------

        unit_price = product["selling_price"]

        gross_amount = (
            unit_price * quantity
        )

        revenue = (
            gross_amount
            * (1 - discount / 100)
        )


        # ----------------------------------------
        # 12. Store transaction
        # ----------------------------------------

        transaction_id = (
            f"T{len(sales_data) + 1:06d}"
        )

        sales_data.append({
            "transaction_id": transaction_id,
            "store_id": store["store_id"],
            "product_id": product["product_id"],
            "customer_id": customer["customer_id"],
            "sale_date": sale_date,
            "quantity": quantity,
            "unit_price": unit_price,
            "discount_percentage": discount,
            "revenue": round(revenue, 2)
        })


# Convert to DataFrame
sales_df = pd.DataFrame(sales_data)

print(
    f"Number of transactions generated: "
    f"{len(sales_df):,}"
)

sales_df.head()
```

```python
# ============================================
# SALES GENERATION — OPTIMIZED VERSION
# ============================================

# Build a fast promotion lookup:
# (date, store_id, product_id) -> discount percentage

promotion_lookup = {}

for _, promo in promotions_df.iterrows():

    active_dates = pd.date_range(
        promo["start_date"],
        promo["end_date"]
    )

    for date in active_dates:

        key = (
            date,
            promo["store_id"],
            promo["product_id"]
        )

        promotion_lookup[key] = promo["discount_percentage"]


print(
    f"Promotion lookup entries: "
    f"{len(promotion_lookup):,}"
)


# --------------------------------------------
# Generate sales
# --------------------------------------------

sales_data = []

date_range = pd.date_range(
    "2026-01-01",
    "2026-06-30"
)

base_daily_transactions = 280


for sale_date in date_range:

    # Determine season
    if sale_date.month <= 2:
        season = "Winter"
    elif sale_date.month == 3:
        season = "Spring"
    else:
        season = "Summer"

    # Day-of-week effect
    daily_multiplier = day_multiplier[
        sale_date.dayofweek
    ]

    # Holiday / event effect
    event_multiplier = 1.0

    if sale_date.month == 1 and sale_date.day == 26:
        event_multiplier = 1.35

    elif sale_date.month == 3 and sale_date.day == 4:
        event_multiplier = 1.40

    elif sale_date.month == 4 and sale_date.day == 1:
        event_multiplier = 1.25

    # Number of daily transaction opportunities
    expected_transactions = int(
        base_daily_transactions
        * daily_multiplier
        * event_multiplier
    )

    num_transactions = max(
        1,
        int(
            np.random.normal(
                expected_transactions,
                expected_transactions * 0.08
            )
        )
    )

    # Generate transactions
    for _ in range(num_transactions):

        # Store
        store = stores_df.iloc[
            np.random.randint(len(stores_df))
        ]

        store_multiplier = store_type_multiplier[
            store["store_type"]
        ]

        size_multiplier = (
            store["store_size_sqft"] / 4500
        )

        # Customer
        customer = customers_df.iloc[
            np.random.randint(len(customers_df))
        ]

        customer_multiplier = customer_segment_multiplier[
            customer["customer_segment"]
        ]

        # Product
        product = products_df.iloc[
            np.random.randint(len(products_df))
        ]

        product_multiplier = product[
            "demand_weight"
        ]

        # Seasonal category effect
        seasonal_multiplier = seasonality[
            season
        ][product["category"]]

        # Promotion lookup
        promotion_key = (
            sale_date,
            store["store_id"],
            product["product_id"]
        )

        discount = promotion_lookup.get(
            promotion_key,
            0
        )

        # Promotion increases demand
        if discount > 0:
            promotion_multiplier = (
                1 + (discount / 100) * 0.50
            )
        else:
            promotion_multiplier = 1.0

        # Combine demand factors
        demand_score = (
            store_multiplier
            * size_multiplier
            * customer_multiplier
            * product_multiplier
            * seasonal_multiplier
            * promotion_multiplier
        )

        # Natural variation
        demand_score *= np.random.uniform(
            0.85,
            1.15
        )

        # Purchase probability
        purchase_probability = min(
            0.95,
            0.30 * demand_score
        )

        if np.random.random() > purchase_probability:
            continue

        # Quantity
        quantity = np.random.choice(
            [1, 2, 3],
            p=[0.75, 0.20, 0.05]
        )

        # Revenue
        unit_price = product["selling_price"]

        gross_amount = (
            unit_price * quantity
        )

        revenue = (
            gross_amount
            * (1 - discount / 100)
        )

        transaction_id = (
            f"T{len(sales_data) + 1:06d}"
        )

        sales_data.append({
            "transaction_id": transaction_id,
            "store_id": store["store_id"],
            "product_id": product["product_id"],
            "customer_id": customer["customer_id"],
            "sale_date": sale_date,
            "quantity": quantity,
            "unit_price": unit_price,
            "discount_percentage": discount,
            "revenue": round(revenue, 2)
        })


# Convert to DataFrame
sales_df = pd.DataFrame(sales_data)

print(
    f"Number of transactions generated: "
    f"{len(sales_df):,}"
)

sales_df.head()
```

```python
# ============================================
# REBUILD PROMOTION LOOKUP
# ============================================

promotion_lookup = {}

for _, promo in promotions_df.iterrows():

    active_dates = pd.date_range(
        promo["start_date"],
        promo["end_date"]
    )

    for date in active_dates:

        key = (
            date,
            promo["store_id"],
            promo["product_id"]
        )

        promotion_lookup[key] = (
            promo["discount_percentage"]
        )

print(
    f"Promotion lookup entries: "
    f"{len(promotion_lookup):,}"
)
```

```python
# ============================================
# FINAL SALES GENERATION
# ============================================

np.random.seed(42)

sales_data = []

date_range = pd.date_range(
    "2026-01-01",
    "2026-06-30"
)

base_daily_transactions = 280

for sale_date in date_range:

    # ----------------------------------------
    # 1. Determine season
    # ----------------------------------------

    if sale_date.month <= 2:
        season = "Winter"
    elif sale_date.month == 3:
        season = "Spring"
    else:
        season = "Summer"

    # ----------------------------------------
    # 2. Day-of-week effect
    # ----------------------------------------

    daily_multiplier = day_multiplier[
        sale_date.dayofweek
    ]

    # ----------------------------------------
    # 3. Holiday / event effect
    # ----------------------------------------

    event_multiplier = 1.0

    if sale_date.month == 1 and sale_date.day == 26:
        event_multiplier = 1.35

    elif sale_date.month == 3 and sale_date.day == 4:
        event_multiplier = 1.40

    elif sale_date.month == 4 and sale_date.day == 1:
        event_multiplier = 1.25

    # ----------------------------------------
    # 4. Daily transaction opportunities
    # ----------------------------------------

    expected_transactions = int(
        base_daily_transactions
        * daily_multiplier
        * event_multiplier
    )

    num_transactions = max(
        1,
        int(
            np.random.normal(
                expected_transactions,
                expected_transactions * 0.08
            )
        )
    )

    # ----------------------------------------
    # 5. Generate transactions
    # ----------------------------------------

    for _ in range(num_transactions):

        # Store
        store = stores_df.iloc[
            np.random.randint(len(stores_df))
        ]

        store_multiplier = store_type_multiplier[
            store["store_type"]
        ]

        size_multiplier = (
            store["store_size_sqft"] / 4500
        )

        # Customer
        customer = customers_df.iloc[
            np.random.randint(len(customers_df))
        ]

        customer_multiplier = (
            customer_segment_multiplier[
                customer["customer_segment"]
            ]
        )

        # Product
        product = products_df.iloc[
            np.random.randint(len(products_df))
        ]

        product_multiplier = product[
            "demand_weight"
        ]

        # ------------------------------------
        # 6. Seasonal category effect
        # ------------------------------------

        seasonal_multiplier = seasonality[
            season
        ][product["category"]]

        # ------------------------------------
        # 7. Promotion effect
        # ------------------------------------

        promotion_key = (
            sale_date,
            store["store_id"],
            product["product_id"]
        )

        discount = promotion_lookup.get(
            promotion_key,
            0
        )

        if discount > 0:

            # Moderate demand uplift
            promotion_multiplier = (
                1 + (discount / 100) * 1.0
            )

        else:

            promotion_multiplier = 1.0

        # ------------------------------------
        # 8. Combine demand factors
        # ------------------------------------

        demand_score = (
            store_multiplier
            * size_multiplier
            * customer_multiplier
            * product_multiplier
            * seasonal_multiplier
            * promotion_multiplier
        )

        # Natural variation
        demand_score *= np.random.uniform(
            0.85,
            1.15
        )

        # ------------------------------------
        # 9. Purchase probability
        # ------------------------------------

        purchase_probability = min(
            0.95,
            0.30 * demand_score
        )

        if np.random.random() > purchase_probability:
            continue

        # ------------------------------------
        # 10. Quantity
        # ------------------------------------

        quantity = np.random.choice(
            [1, 2, 3],
            p=[0.75, 0.20, 0.05]
        )

        # ------------------------------------
        # 11. Revenue
        # ------------------------------------

        unit_price = product["selling_price"]

        gross_amount = (
            unit_price * quantity
        )

        revenue = (
            gross_amount
            * (1 - discount / 100)
        )

        # ------------------------------------
        # 12. Store transaction
        # ------------------------------------

        transaction_id = (
            f"T{len(sales_data) + 1:06d}"
        )

        sales_data.append({
            "transaction_id": transaction_id,
            "store_id": store["store_id"],
            "product_id": product["product_id"],
            "customer_id": customer["customer_id"],
            "sale_date": sale_date,
            "quantity": quantity,
            "unit_price": unit_price,
            "discount_percentage": discount,
            "revenue": round(revenue, 2)
        })


# Convert to DataFrame
sales_df = pd.DataFrame(sales_data)

print(
    f"Number of transactions generated: "
    f"{len(sales_df):,}"
)

sales_df.head()
```

```python
# ============================================
# PREPARE SALES FOR INVENTORY GENERATION
# ============================================

daily_sales = (
    sales_df
    .groupby(
        ["sale_date", "store_id", "product_id"],
        as_index=False
    )
    .agg(
        units_sold=("quantity", "sum")
    )
)

print(
    "Daily store-product combinations:",
    len(daily_sales)
)

daily_sales.head()
```

```python
# ============================================
# INVENTORY GENERATION
# ============================================

inventory_data = []

# Track previous day's closing stock
stock_tracker = {}

date_range = pd.date_range(
    "2026-01-01",
    "2026-06-30"
)

# Fast lookup for units sold
sales_lookup = (
    daily_sales
    .set_index(
        ["sale_date", "store_id", "product_id"]
    )["units_sold"]
    .to_dict()
)

inventory_number = 1

for sale_date in date_range:

    for store_id in stores_df["store_id"]:

        for product_id in products_df["product_id"]:

            key = (
                sale_date,
                store_id,
                product_id
            )

            # --------------------------------
            # Previous closing stock
            # --------------------------------

            previous_key = (
                store_id,
                product_id
            )

            opening_stock = stock_tracker.get(
                previous_key,
                np.random.randint(15, 40)
            )

            # --------------------------------
            # Units sold
            # --------------------------------

            units_sold = sales_lookup.get(
                key,
                0
            )

            # --------------------------------
            # Normal replenishment
            # --------------------------------

            units_received = 0

            # Replenish when stock is getting low
            if opening_stock <= max(
                5,
                units_sold
            ):

                units_received = np.random.randint(
                    10,
                    35
                )

            # --------------------------------
            # S007 stockout scenario
            # --------------------------------

            if (
                store_id == "S007"
                and sale_date.month == 6
            ):

                # Deliberately weak replenishment
                units_received = np.random.randint(
                    0,
                    5
                )

            # --------------------------------
            # Calculate closing stock
            # --------------------------------

            closing_stock = max(
                0,
                opening_stock
                + units_received
                - units_sold
            )

            # --------------------------------
            # Stockout flag
            # --------------------------------

            stockout_flag = (
                closing_stock == 0
            )

            # --------------------------------
            # Store result
            # --------------------------------

            inventory_data.append({
                "inventory_id":
                    f"INV{inventory_number:07d}",

                "store_id":
                    store_id,

                "product_id":
                    product_id,

                "date":
                    sale_date,

                "opening_stock":
                    opening_stock,

                "units_received":
                    units_received,

                "units_sold":
                    units_sold,

                "closing_stock":
                    closing_stock,

                "stockout_flag":
                    stockout_flag
            })

            # Update tracker
            stock_tracker[
                previous_key
            ] = closing_stock

            inventory_number += 1


inventory_df = pd.DataFrame(
    inventory_data
)

print(
    f"Inventory records generated: "
    f"{len(inventory_df):,}"
)

inventory_df.head()
```
